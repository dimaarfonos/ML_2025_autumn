#### Поскольку у нас нет заранее размеченных данных (мы не знаем наверняка, какой день к какому "типу" относится с точки зрения энергосети), мы будем использовать Unsupervised Learning (обучение без учителя), а именно алгоритм K-Means.

Алгоритм сам найдет дни с похожим графиком потребления и объединит их в кластеры.



Готовим данные: Превращает таблицу в матрицу, где строка — это одна дата, а столбцы (0–23) — потребление по часам.

Кластеризация: Разбивает все дни на 3 типа (кластера).

Визуализация: Рисует средний график для каждого кластера (чтобы вы поняли: "Ага, Кластер 0 — это когда никого нет дома, а Кластер 2 — это выходные").

Анализ: Считает ночное потребление для каждого кластера (для подбора батареи).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# --- 1. ЗАГРУЗКА И ПОДГОТОВКА ДАННЫХ ---
# Загружаем файл
df = pd.read_csv('power_usage_2016_to_2020.csv')

# Преобразуем дату в правильный формат
df['StartDate'] = pd.to_datetime(df['StartDate'])

# Выделяем саму дату (без времени) и час
df['Date'] = df['StartDate'].dt.date
df['Hour'] = df['StartDate'].dt.hour

# Создаем сводную таблицу (Pivot Table)
# Строки = Дни, Столбцы = Часы (0..23), Значения = Потребление (kWh)
# Это и есть наши "признаки" для ML
daily_profiles = df.pivot(index='Date', columns='Hour', values='Value (kWh)')

# Удаляем дни, где есть пропуски (если такие есть)
daily_profiles = daily_profiles.dropna()

print(f"Данные подготовлены. Всего дней для анализа: {daily_profiles.shape[0]}")

# --- 2. MACHINE LEARNING: K-MEANS ---

# Нормализуем данные (StandardScaler), чтобы алгоритм работал корректно
scaler = StandardScaler()
X_scaled = scaler.fit_transform(daily_profiles.values)

# Запускаем K-Means
# n_clusters=3 — попробуем найти 3 типичных сценария (например: "Экономный", "Обычный", "Высокая нагрузка")
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

# Добавляем метку кластера обратно к нашим данным
daily_profiles['Cluster'] = clusters

# --- 3. ВИЗУАЛИЗАЦИЯ РЕЗУЛЬТАТОВ ---

# Считаем средний профиль потребления для каждого кластера
cluster_means = daily_profiles.groupby('Cluster').mean()

plt.figure(figsize=(12, 6))
# Рисуем графики
for cluster_id in cluster_means.index:
    plt.plot(cluster_means.columns, cluster_means.loc[cluster_id], 
             label=f'Кластер {cluster_id}', linewidth=2)

plt.title('Средние профили потребления энергии по кластерам')
plt.xlabel('Час суток (0-23)')
plt.ylabel('Среднее потребление (кВт*ч)')
plt.xticks(range(0, 24))
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

# --- 4. АНАЛИЗ ДЛЯ СОЛНЕЧНЫХ БАТАРЕЙ ---
# Задача: найти кластер с самым высоким ночным потреблением
# Ночь по заданию: с 18:00 до 08:00.

# Часы, которые относятся к "ночи" (18..23 и 0..7)
night_hours = [18, 19, 20, 21, 22, 23, 0, 1, 2, 3, 4, 5, 6, 7]

# Считаем сумму потребления в эти часы для каждого кластера
cluster_means['Night_Consumption'] = cluster_means[night_hours].sum(axis=1)

print("\n--- Результаты анализа для подбора аккумулятора ---")
for cluster_id, row in cluster_means.iterrows():
    print(f"Кластер {cluster_id}: Потребление ночью (18:00-08:00) = {row['Night_Consumption']:.2f} кВт*ч")

max_night_usage = cluster_means['Night_Consumption'].max()
print(f"\n>>> РЕКОМЕНДАЦИЯ: Для полной автономности батарея должна иметь емкость не менее {max_night_usage:.2f} кВт*ч")

# --- 5. ПРОВЕРКА ГИПОТЕЗ (Связь с COVID/Выходными) ---
# Присоединим исходные метки (notes), чтобы понять суть кластеров
df_meta = df[['StartDate', 'notes']].copy()
df_meta['Date'] = df_meta['StartDate'].dt.date
df_meta = df_meta.drop_duplicates(subset=['Date']).set_index('Date')

# Объединяем
analysis_df = daily_profiles.join(df_meta[['notes']])

print("\n--- Как распределились дни по кластерам? ---")
print(pd.crosstab(analysis_df['notes'], analysis_df['Cluster']))
```

 3 линии.

Одна линия будет ниже всех — это "Режим экономии" (скорее всего, дни отпуска или когда никого нет дома).


Одна линия будет выше всех или иметь пики днем — это "Активный режим" (выходные или удаленка в COVID).

Третья — промежуточная.

Таблица (Crosstab): В конце код выведет таблицу. Посмотрите на пересечение колонок Cluster и строк notes.

Если вы увидите, что COVID_lockdown почти целиком попал в один кластер (например, в Кластер 2), вы сможете сделать вывод: "Пандемия сместила потребление в сторону режима [такого-то], увеличив нагрузку днем"

Батарея: Код автоматически посчитает, сколько энергии тратится ночью в самом "тяжелом" режиме. Это и есть ответ на инженерную задачу проекта.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# --- 1. ЗАГРУЗКА ---
df = pd.read_csv('power_usage_2016_to_2020.csv')
df['StartDate'] = pd.to_datetime(df['StartDate'])
df = df.sort_values('StartDate').reset_index(drop=True)

# --- 2. ГЕНЕРАЦИЯ ПРИЗНАКОВ (Feature Engineering) ---
# Модель не понимает даты, ей нужны числа
df['Hour'] = df['StartDate'].dt.hour
df['DayOfWeek'] = df['StartDate'].dt.dayofweek
df['Month'] = df['StartDate'].dt.month
df['is_weekend'] = df['DayOfWeek'].isin([5, 6]).astype(int)

# Самый сильный признак: "Сколько потребляли вчера в это же время?"
# (Lag features)
df['lag_24h'] = df['Value (kWh)'].shift(24) # 24 часа назад
df['lag_168h'] = df['Value (kWh)'].shift(24 * 7) # Неделю назад

# Удаляем пустые строки, которые появились из-за сдвига (первая неделя)
df = df.dropna()

# --- 3. ПОДГОТОВКА ML ---
# Используем данные до 2020 года для обучения, а 2020 год - для теста
# Это честная проверка "в будущем"
split_date = pd.to_datetime('2020-01-01')

train = df[df['StartDate'] < split_date]
test = df[df['StartDate'] >= split_date]

features = ['Hour', 'DayOfWeek', 'Month', 'is_weekend', 'lag_24h', 'lag_168h']
target = 'Value (kWh)'

X_train = train[features]
y_train = train[target]
X_test = test[features]
y_test = test[target]

print(f"Обучающая выборка: {X_train.shape[0]} часов")
print(f"Тестовая выборка: {X_test.shape[0]} часов")

# --- 4. ОБУЧЕНИЕ МОДЕЛИ (Random Forest) ---
# Random Forest отлично подходит для таких задач
rf = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)

# --- 5. ПРОВЕРКА ---
predictions = rf.predict(X_test)

# Оценка точности (R2 score: 1.0 = идеально, 0.0 = плохо)
score = r2_score(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))

print(f"Точность модели (R2): {score:.2f}")
print(f"Средняя ошибка (RMSE): {rmse:.2f} кВт*ч")

# --- 6. ВИЗУАЛИЗАЦИЯ (Магия для презентации) ---
# Построим график за одну неделю из теста, чтобы было видно детально
week_start = 0
week_end = 168 # 24 * 7 часов

plt.figure(figsize=(15, 6))
plt.plot(test['StartDate'].iloc[week_start:week_end], 
         y_test.iloc[week_start:week_end], 
         label='Реальное потребление', color='blue', alpha=0.6)

plt.plot(test['StartDate'].iloc[week_start:week_end], 
         predictions[week_start:week_end], 
         label='Прогноз ИИ', color='orange', linestyle='--', linewidth=2)

plt.title('Тест модели: Реальность vs Прогноз (одна неделя в 2020)')
plt.ylabel('Потребление (кВт*ч)')
plt.legend()
plt.grid(True)
plt.show()

# --- 7. ВАЖНОСТЬ ПРИЗНАКОВ ---
# Что больше всего влияет на потребление?
importances = pd.Series(rf.feature_importances_, index=features).sort_values(ascending=False)
print("\nЧто влияет на потребление (топ факторов):")
print(importances)
```
